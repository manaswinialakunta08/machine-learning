```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder #converts categorical into numerical
from sklearn.preprocessing import StandardScaler #Performs Feature scaling
from sklearn.impute import SimpleImputer #To handle missingvalue
from sklearn.model_selection import train_test_split
df=pd.read_csv("/content/user_datacsv__2026_06_12_12_31_57 (2).csv")
print(df)

```

```python
#Handling MissingValues
print(df.isnull().sum())
```

```python
#creating imputer to calculate missingvalues
imputer = SimpleImputer(strategy='mean')
df[['Age', 'Est_salary']] = imputer.fit_transform(
    df[['Age', 'Est_salary']]
)
print(df)
```

```python
label_encoder=LabelEncoder()
df['Gender']=label_encoder.fit_transform(df['Gender'])
print(df)
```

```python
#Feature Scaling
scaler=StandardScaler()
df[['Age','Est_salary']]=scaler.fit_transform(df[['Age','Est_salary']])
print(df)
```

```python
#creating independent and dependent variables
x=['Gender','Age','Est_salary']
print(x)
y=['Purchased']
print(y)
```
