```python
import pandas as pd
tennis=pd.read_csv("/content/Tennisdataset.csv")
print(tennis)
```

```python
tennis["windy"]=tennis["windy"].astype(str)
```

```python
ct_outlook=pd.crosstab(
    tennis["outlook"],
    tennis["play"],
    margins=True
)
print(ct_outlook)
```

```python
ct_temp=pd.crosstab(
    tennis["temp"],
    tennis["play"],
    margins=True
)
print(ct_temp)
```

```python
ct_humidity=pd.crosstab(
    tennis["humidity"],
    tennis["play"],
    margins=True
)
print(ct_humidity)
```

```python
ct_windy=pd.crosstab(
    tennis["windy"],
    tennis["play"],
    margins=True
)
print(ct_windy)
```

```python
ct_play=pd.crosstab(
    tennis["play"],
    columns="count",
    margins=True
)
print(ct_play)
```

```python
#Prediction Function
def predict(outlook,temp,humidity,windy):
  #calculate p(yes)
  p_yes=ct_play.loc["yes","count"]/ct_play.loc["All","count"]
  #calculate conditionalprobabilities
  p_outlook_yes=ct_outlook.loc[outlook,"yes"]/ct_outlook.loc["All","yes"]
  p_temp_yes=ct_temp.loc[temp,"yes"]/ct_temp.loc["All","yes"]
  p_humidity_yes=ct_humidity.loc[humidity,"yes"]/ct_humidity.loc["All","yes"]
  p_windy_yes=ct_windy.loc[windy,"yes"]/ct_windy.loc["All","yes"]
  #calculate prediction
  p_yes_given_x=p_yes*p_outlook_yes*p_temp_yes*p_humidity_yes*p_windy_yes
  #calculate p(no)
  p_no=ct_play.loc["no","count"]/ct_play.loc["All","count"]
  #calculate conditionalprobabilities
  p_outlook_no=ct_outlook.loc[outlook,"no"]/ct_outlook.loc["All","no"]
  p_temp_no=ct_temp.loc[temp,"no"]/ct_temp.loc["All","no"]
  p_humidity_no = ct_humidity.loc[humidity, "no"] / ct_humidity.loc["All", "no"]
  p_windy_no = ct_windy.loc[windy, "no"] / ct_windy.loc["All", "no"]
  #calculate prediction
  p_no_given_x=p_no*p_outlook_no*p_temp_no*p_humidity_no*p_windy_no
  #Final Prediction
  if p_yes_given_x > p_no_given_x:
    return "yes"
  else:
    return "no"

test_samples = [
    ("sunny", "hot", "high", "True", "no"),
    ("overcast", "cool", "normal", "False", "yes"),
    ("rainy", "mild", "high", "False", "yes")
]
correct = 0

for outlook, temp, humidity, windy, actual in test_samples:

    prediction = predict(
        outlook,
        temp,
        humidity,
        windy
    )

    print(
        f"Sample: ({outlook}, {temp}, {humidity}, {windy}) "
        f"-> Predicted: {prediction}, Actual: {actual}"
    )

    if prediction == actual:
        correct += 1


# Accuracy Calculation

accuracy = (correct / len(test_samples)) * 100

print(f"\nAccuracy: {accuracy:.2f}%")




```
